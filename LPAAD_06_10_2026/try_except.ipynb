#### TRY e EXCEPT



- TRY:

TRY de TENTAR: vai testar o pedaço do código que pode gerar uma exceção

- EXCEPT:

Em conjunto com o TRY, ele é acionado quando ocorre a exceção tratada no bloco de código do TRY.

- ELSE??:

Sim, Else também pode ser usado por aqui, quando TRY não aciona o Except, o ELSE é executado.

- Finally:
Basicamente, ele é acionado independente se o TRY,Except ou else for acionado ou não.


```python
# print('para de dar erro')
try:
  idade = int(input("digite uma idade:"))

except ValueError:
  print('Digite VALOR NUMÉRICO')
finally:
  print('\nOk programa finalizada.')
```

##### Prática 1 : Operação Bancária

- Usuário digita o valor do saque de 100 á 5000
- Valor do saque não deve extrapolar o limite restante de crédito. valor inicial: 8000
- Usuário recebe opções de Parcelas(Nº) válidas
- Usuáro seleciona Juros Simples ou Compostos e retorna o valor do montante final do Saque inicial
- Principal: O código deve Validar valores dados pelo usuário e notificar possiveis erros/exceções em todo processo do saque

```python
limite_saque = 8000

# executando o saque dentro do limite
while True:
  try:
    valor_saque = float(input('\ndigite um valor(R$) entre R$100 Á R$5000:\n'))
  except ValueError:
    print('ERRO: Digite um valor NUMÉRICO DE 100 Á 5000')
    print('---'*5)
    break

  else:
    if not valor_saque > 100 or valor_saque > 5000:
      print('ERRO: fora do limite do saldo mané')
      print('---'*5)
      break
    if limite_saque < 500:
      print('Limite de crédito abaixo de R$500 - SAQUE CANCELADO')
      break

    # limite disponivel agora
    print('-- saque concluido --')
    limite_saque = limite_saque - valor_saque
    print(f'limite atualizado: R$ {limite_saque}')

    # Parcelando o saque??
    parcelas = []

    # preciso verificar as possibilidades de parcela
    print('Opções de Parcela disponiveis:')
    for p in range(6):
      numero = p
      p+=1
      valor_parcela = valor_saque / p
      if valor_parcela > 50:
        parcelas.append(p)
        print(f'DIGITE --{numero}-- para parcela em {p} x vezes')

    print('---'*5)
    # caso o usuario digite valores NÃO-NUMÉIRCOS
    try:
      opcao_parcela = int(input('Digite a opção desejável:'))
    except ValueError:
      print('ERRO: Digite um valor NUMÉRICO MDS')
      print('---'*5)
      break
    # Outro aqui, pode ocorrer erro na condição do IF caso o valor esteja fora do index
    try:
      if not parcelas[opcao_parcela] in parcelas:
        break
    except:
      print('ERRO: Digite um valor VÁLIDO ENTRE AS OPÇÕES MDS')
      print('---'*5)
      break


    print('---'*5)
    # ESCOLHENDO a opção de JUROS?
    print('DIGITE (1) para Juros simples')
    print('DIGITE (2) para Juros Compostos')
    opcao_juros = int(input())

    if opcao_juros == 1:
      juros = valor_saque * 0.02 * parcelas[opcao_parcela]
      valor_c_juros = valor_saque + juros
      print(f'Montante Final com Juros simples: R${valor_c_juros}')

    elif opcao_juros == 2:
      valor_c_juros = valor_saque * (1 + 0.02) ** parcelas[opcao_parcela]
      print(f'Montante Final com Juros compostos: R${valor_c_juros}')


  finally:
    print('\nOperação finalizada')
```
