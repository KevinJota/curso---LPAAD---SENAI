### ENUMERATE

funciona como contador/informar posicao basicamente... prorvavelmente.
basicamente pega uma lista de valores e ENUMERA ELES(wow)


```python
produtos = ['celular','Notebook','TECLADO']

# start seleciona apartir de qual numeração começa a contagem do ENUMERATE
# gambiarra do indice+1 para numerar cabou

for numero, produto in enumerate(produtos, start=1):
    print(numero,produto)
```

```python
vendedores = [
    {"nome": "Ana", "Vendas": 12500},
    {"nome": "Carlos", "Vendas": 9800},
    {"nome": "Marina", "Vendas": 15700},
    {"nome": "João", "Vendas": 11300}
]

for numero, nome in enumerate(vendedores, start=1):
    print(numero,nome)
```
