# DATETIME

modulo/biblioteca usada pra armazenar dados temporais(dia,ano,mes,hora,segundo,minuto)

```python
from datetime import datetime

AGORA = datetime.now()
print(AGORA) # Basicamente, pega a data e hora atual registrados pela sua máquina/dispositivo
```

- Formatando a data com STRFTIME:

```python
data_formatada = AGORA.strftime("%d/%m/%Y")

print(data_formatada)
print(type(data_formatada)) # UE,virou string quando faz isso
```

- Formatando a data e HORARIO também com STRFTIME:

```python
data_formatada_2 = AGORA.strftime("%d/%m/%Y %H:%M") # colocaria :%S para formatar e exibir os SEGUNDOS

print(data_formatada_2)
print(type(data_formatada_2)) 
# virou string tbm.... mas tá formatada
# é interesante trabalhar com isso ao exibir em RESULTADOS FINAIS/frontend DA VIDA ai
```

- Convertendo Texto para data: STRPTIME() - string parse time

```python
data_texto = "21/09/2026"
# confirmando o tipo....
print(type(data_texto))
print(data_texto)

print('***'*5)
data = datetime.strptime(data_texto,"%d/%m/%Y")

#Apos CONVERSÃO
print(type(data))
print(data)
```

- operacoes matematica com DATAS

```python
#eba...

data_pedido = "10/09/2026"
data_entrega = "15/09/2026"

data1 = datetime.strptime(data_pedido,"%d/%m/%Y")
data2 = datetime.strptime(data_entrega,"%d/%m/%Y")

# dataa = datetime(2026, 9, 17)
# print(dataa)

print(data2)
print(data1)
# formatando a data na saida usando o STRF DE NOVO(data2 é datetime ainda)
print(f'data formatada direto no print: {data2.strftime("%d/%m/%Y")}')

diferenca = data2 - data1

print('#'*15)

print(diferenca)
print(type(diferenca))

print('#'*15)

print(type(diferenca.days))
# pra retirar o time e deixar só os dias de diferenca...(INT basicamente)
print(f'tempo de entrega estipulado: {diferenca.days} dias')

#como raios tiro essa cotação de 00:00:00??
# o tipo de dado aqui é datetime, ou seja: date + TIME(HORA/MINUTO/SEGUNDO)
# sugestão de meelhoria, dava pra fazer uma FUNÇÃO pra formatar data ai....
```
