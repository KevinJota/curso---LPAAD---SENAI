### PROBLEMA 1

```python
equipeA = [42, 38, 45, 41, 39, 44, 38]
equipeB = [30, 52, 35, 48, 29, 55, 38]

```

```python
def calcular_media(PARAMETRO):
    media = sum(PARAMETRO) / len(PARAMETRO)
    return media
```

```python
resA = calcular_media(equipeA)
resB = calcular_media(equipeB)

print(f'média da equipe A: {resA}')
print(f'média da equipe B: {resB}')
```

```python
medianaA = statistics.median(equipeA)
medianaB = statistics.median(equipeB)

print(f'mediana da equipe A: {medianaA}')
print(f'mediana da equipe B: {medianaB}')
```

```python
moda_a = statistics.multimode(equipeA)
moda_b = statistics.multimode(equipeB)

print(f'MODA da equipe A: {moda_a}')
print(f'MODA da equipe B: {moda_b}')
```

```python
amplitude_a = max(equipeA) - min(equipeA)
amplitude_b = max(equipeB) - min(equipeB)

print(f'amplitude da equipe A: {amplitude_a}')
print(f'amplitude da equipe B: {amplitude_b}')
```

### PROBLEMA 3

```python
tempos = [ 48, 32, 55, 41, 37, 62, 29, 45, 51, 35, 70]
```

```python
def calcular_valor_quartil(dados, posicao):
    posicao_inteira = int(posicao)
    if posicao == posicao_inteira:
        return dados[posicao_inteira - 1]
    
    parte_decimal = posicao - posicao_inteira
    valor_inferior = dados[posicao_inteira - 1]
    valor_superior = dados[posicao_inteira]

    diferenca = valor_superior - valor_inferior
    acrescimo = diferenca * parte_decimal

    return valor_inferior + acrescimo

vendas = [42,18,35,27,50,31,22]


def calcula_quartil(lista):
    dados = sorted(lista)
    valor = len(dados)
    PQ1_completo = (valor + 1) / 4 # para pegar a posição completa
    PQ2_completo = (valor + 1) / (4 / 2) # para pegar a posição completa
    PQ3_completo = (3 * (valor / 1)) / 4 # para pegar a posição completa

    q1 = calcular_valor_quartil(dados, PQ1_completo)
    q2 = calcular_valor_quartil(dados, PQ2_completo)
    q3 = calcular_valor_quartil(dados, PQ3_completo)
    return q1, q2, q3

def calcular_intervalo_interquartil(lista):

    q1, _, q3 = calcula_quartil(lista)
    return q3 - q1
```

```python
media = statistics.mean(tempos)
mediana = statistics.median(tempos)

q1, q2, q3 = calcula_quartil(tempos)

amplitude = max(tempos) - min(tempos)
desvio_padrao = statistics.stdev(tempos)
variancia = statistics.variance(tempos)

iqr = calcular_intervalo_interquartil(tempos)

print(f' tempo médio das entregas em minutos: {media:.2f}')
print(f' tempo mediano geral das entregas: {mediana:.0f}')
print('---------------------------------------------------')

print(f'Amplitude do tempo de entrega: {amplitude}')
print(f'desvio padrão do tempo de entrega: {desvio_padrao:.2f}')
print(f'variancia do tempo de entrega: {variancia:.2f}')
print(f'IQR: {iqr}')

print('---------------------------------------------------')
print(f'Q1: {q1}')
print(f'Q2: {q2}')
print(f'Q3: {q3}')
```
